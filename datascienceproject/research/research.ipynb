```python
## Explanation is for @ensure_annotation.
def get_product(x: int, y: int) -> int:
    return x * y

result = get_product(4, 5)
print(f"The product of two number is: {result}")
print(result)

result1 = get_product(4, "6")
print(f"The product of two numbers is: {result1}") # Output: '6666'


# After Implementations of @ensure_annotations
# It can handle such kinds of situations. 
from ensure import ensure_annotations

@ensure_annotations
def get_product(x: int, y: int) -> int:
    return x * y

result3 = get_product(4, '5')
print(f"The product of two number is: {result3}") # Output: 20
```
